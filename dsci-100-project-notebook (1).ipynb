```python
import altair as alt

alt.Chart(tidy_data).mark_bar().encode(
    alt.X("reg_wins_pct:Q", title="Regular Season Win Percentage"),
    alt.Y("count()", title="Number of teams")
).properties(title="Distribution of Regular Season Win Percentage")
```

```python
props = (tidy_data["bid"].value_counts(normalize=True)
         .rename_axis("bid").reset_index(name="proportion"))
alt.Chart(props).mark_bar().encode(
    alt.X("bid:N", title="Bid Type"),
    alt.Y("proportion:Q", title="Proportion of Teams")
).properties(title="Proportion of Teams by Bid Type")
```

```python
alt.Chart(tidy_data).mark_circle(opacity=0.5).encode(
    alt.X("reg_wins_pct:Q", title="Regular Season Win Percentage"),
    alt.Y("seed:Q", title="Tournament Seed"),
    alt.Color("bid:N", title="Bid Type")
).properties(title="Win Percentage VS Seed by Bid Type")
```

This scatterplot shows regular season win percentage against tournament seed, coloured by bid type. It is relevant because our research question asks whether a team's record can predict its bid type. If automatic and at-large teams cluster in different areas, these variables should help classify them. Heavy overlap would suggest the prediction will be harder.

```python
alt.Chart(tidy_data).mark_boxplot().encode(
    alt.X("bid:N", title = "Bid Type"),
    alt.Y("reg_wins_pct:Q", title="Regular Season Win Percentage")
).properties(title="Win Percentage by Bid Type")
```

This boxplot compares regular season win percentage for automatic and at-large teams. It shows whether the two groups differ in typical win percentage and spread. A clear gap between the medians suggests win percentage is a useful predictor of bid type, while heavily overlapping boxes suggest it is weak on its own.
